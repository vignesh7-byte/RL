```python

gamma = 0.9
theta = 0.000001

states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

transitions = {
    "S0": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S0", -1),
        "Right": ("S0", -1)
    },
    "S1": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S1", -1),
        "Right": ("G", 100)
    }
}

V = {
    "S0": 0.0,
    "S1": 0.0,
    "G": 0.0
}

print("MDP INFORMATION")
print("----------------------")
print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)

print("\nTRANSITION DETAILS")
print("----------------------")

for s in states:
    for a in actions:
        next_state, reward = transitions[s][a]
        print(s, a, "->", next_state, "Reward:", reward)

print("\nINITIAL STATE VALUES")
print("----------------------")
print("V(S0) =", V["S0"])
print("V(S1) =", V["S1"])

print("\nVALUE ITERATION")
print("----------------------")

iteration = 0

while True:

    new_V = V.copy()

    for s in states:

        q_values = []

        for a in actions:
            next_state, reward = transitions[s][a]

            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[s] = max(q_values)

    delta = max(
        abs(new_V[s] - V[s]) for s in states
    )

    V = new_V
    iteration += 1

    print(
        "Iteration", iteration,
        "V(S0) =", round(V["S0"], 2),
        "V(S1) =", round(V["S1"], 2),
        "Delta =", round(delta, 6)
    )

    if delta < theta:
        break

print("\nFINAL OPTIMAL STATE VALUES")
print("----------------------")

for s in states:
    print(s, "=", round(V[s], 2))

print("G = 0")

print("\nFINAL ACTION VALUES")
print("----------------------")

Q = {}

for s in states:

    Q[s] = {}

    for a in actions:

        next_state, reward = transitions[s][a]

        Q[s][a] = reward + gamma * V[next_state]

        print(
            "Q(", s, ",", a, ") =",
            round(Q[s][a], 2)
        )

print("\nOPTIMAL POLICY")
print("----------------------")

for s in states:

    best_value = max(Q[s].values())

    best_actions = [
        a for a in actions
        if abs(Q[s][a] - best_value) < 0.0000001
    ]

    print(s, "->", best_actions)

print("\nOPTIMAL PATH")
print("----------------------")
print("S0 -> S1 -> G")
```
