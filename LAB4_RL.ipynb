```python
import math

machines = ['A', 'B', 'C']
N = [0, 0, 0]
R = [0, 0, 0]
UCB = [0, 0, 0]
Q = [0.0, 0.0, 0.0]

N_play = 10
total_reward = 0

print("Upper Confidence Bound (UCB) Multi-Armed Bandit")
print("Machines:", machines)
print()

for play in range(1, N_play + 1):
    for i in range(len(machines)):
        if N[i] == 0:
            UCB[i] = float('inf')
        else:
            UCB[i] = Q[i] + math.sqrt((2 * math.log(play)) / N[i])

    selected = UCB.index(max(UCB))

    print("Play", play)
    print("UCB values:", UCB)
    print("Selected machine:", machines[selected])

    reward = float(input("Enter reward: "))

    N[selected] += 1
    R[selected] += reward
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    total_reward += reward

    print("Reward received:", reward)
    print("Updated Q-value:", Q[selected])
    print()

for i in range(len(machines)):
    if N[i] == 0:
        UCB[i] = float('inf')
    else:
        UCB[i] = Q[i] + math.sqrt((2 * math.log(N_play)) / N[i])

print("Final Results")
print()

print("Machine\tN\tR\tUCB\t\tQ")
for i in range(len(machines)):
    print(machines[i], "\t", N[i], "\t", R[i], "\t", round(UCB[i], 4), "\t\t", round(Q[i], 4))

print()
print("Total Reward =", total_reward)
```
