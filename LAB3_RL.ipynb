```python
import random

machines = ['A', 'B', 'C']
epsilon = 0.2
n_plays = 10

N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]

total_reward = 0

for play in range(1, n_plays + 1):
    r = random.random()

    highest_q = max(Q)
    highest_indices = [i for i in range(len(machines)) if Q[i] == highest_q]

    if r < epsilon:
        available_machines = [i for i in range(len(machines)) if i not in highest_indices]
        selected = random.choice(available_machines)
        decision = "Exploration"
    else:
        selected = random.choice(highest_indices)
        decision = "Exploitation"

    print("\nPlay", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Play with machine", machines[selected])

    reward = float(input("Enter reward: "))

    N[selected] += 1
    R[selected] += reward
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    total_reward += reward

    print("N values:", N)
    print("R values:", R)
    print("Q values:", [round(x, 2) for x in Q])

print("\nFinal Results")
print("-------------------------")

for i in range(len(machines)):
    print("Machine", machines[i])
    print("Participation Count N =", N[i])
    print("Total Reward R =", R[i])
    print("Estimated Q =", round(Q[i], 2))
    print()

print("Total Reward after 10 plays =", total_reward)
```
