```python

gamma = 0.9
theta = 0.000001

states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

transitions = {
    "S0": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S0", -1),
        "Right": ("S0", -1)
    },
    "S1": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S1", -1),
        "Right": ("G", 100)
    }
}

V = {"S0": 0.0, "S1": 0.0, "G": 0.0}

policy = {
    "S0": "Up",
    "S1": "Up"
}

print("MDP INFORMATION")
print("----------------------")
print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)
print("Goal Reward: 100")
print("Normal Reward: -1")
print("Initial Policy:", policy)

print("\nTRANSITION DETAILS")
print("----------------------")

for s in states:
    for a in actions:
        next_state, reward = transitions[s][a]
        print(s, a, "->", next_state, "Reward:", reward)

iteration = 0

while True:

    iteration += 1

    print("\nPOLICY ITERATION:", iteration)

    print("Current Policy:", policy)

    eval_iteration = 0

    while True:

        new_V = V.copy()

        for s in states:
            a = policy[s]
            next_state, reward = transitions[s][a]

            new_V[s] = reward + gamma * V[next_state]

        delta = max(
            abs(new_V[s] - V[s]) for s in states
        )

        V = new_V
        eval_iteration += 1

        print(
            "Evaluation", eval_iteration,
            "V(S0) =", round(V["S0"], 6),
            "V(S1) =", round(V["S1"], 6),
            "Delta =", round(delta, 6)
        )

        if delta < theta:
            break

    new_policy = {}

    Q = {}

    for s in states:

        Q[s] = {}

        for a in actions:

            next_state, reward = transitions[s][a]

            Q[s][a] = reward + gamma * V[next_state]

        best_value = max(Q[s].values())

        best_action = [
            a for a in actions
            if abs(Q[s][a] - best_value) < 0.0000001
        ]

        new_policy[s] = best_action[0]

    print("Improved Policy:", new_policy)

    if new_policy == policy:
        print("Policy is Stable")
        break

    policy = new_policy

print("\nFINAL OPTIMAL STATE VALUES")
print("----------------------")

for s in states:
    print(s, "=", round(V[s], 2))

print("G = 0")

print("\nFINAL ACTION VALUES")
print("----------------------")

for s in states:
    for a in actions:
        print(
            "Q(", s, ",", a, ") =",
            round(Q[s][a], 2)
        )

print("\nOPTIMAL POLICY")
print("----------------------")

for s in states:
    print(s, "->", policy[s])

print("\nBELLMAN OPTIMALITY VERIFICATION")
print("----------------------")

for s in states:

    max_q = max(Q[s].values())

    print(
        s,
        "V* =", round(V[s], 2),
        "Max Q* =", round(max_q, 2),
        "Verified =", abs(V[s] - max_q) < theta
    )
```
