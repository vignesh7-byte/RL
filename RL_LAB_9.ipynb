```python

gamma = 0.9

V_S0 = 71.819 / 0.9271
V_S1 = 89.9 + 0.09 * V_S0

Q_S0_Down = -1 + gamma * V_S1
Q_S0_Right = -1
Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V_S0

check_S0 = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
check_S1 = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

print("ASSIGNMENT 1 RESULTS")
print("-------------------------")

print("State Values:")
print("V(S0) =", round(V_S0, 2))
print("V(S1) =", round(V_S1, 2))

print("\nAction Values:")
print("Q(S0, Down) =", round(Q_S0_Down, 2))
print("Q(S0, Right) =", round(Q_S0_Right, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))
print("Q(S1, Up) =", round(Q_S1_Up, 2))

print("\nVerification:")
print("Calculated V(S0) =", round(check_S0, 2))
print("Calculated V(S1) =", round(check_S1, 2))
```

```python

gamma = 0.9
theta = 0.000001

V_S0 = 0.0
V_S1 = 0.0

iterations = 0

while True:

    old_S0 = V_S0
    old_S1 = V_S1

    new_S0 = 0.1 * (-1 + gamma * 0) + \
             0.9 * (-1 + gamma * old_S1)

    new_S1 = 0.9 * (100 + gamma * 0) + \
             0.1 * (-1 + gamma * old_S0)

    delta = max(
        abs(new_S0 - old_S0),
        abs(new_S1 - old_S1)
    )

    V_S0 = new_S0
    V_S1 = new_S1

    iterations += 1

    if delta < theta:
        break

Q_S0_Down = 0.1 * (-1 + gamma * 0) + \
            0.9 * (-1 + gamma * V_S1)

Q_S0_Right = -1 + gamma * 0

Q_S1_Right = 0.9 * (100 + gamma * 0) + \
             0.1 * (-1 + gamma * V_S0)

Q_S1_Up = -1 + gamma * V_S0

print("ASSIGNMENT 2 RESULTS")
print("-------------------------")

print("Number of Iterations:", iterations)

print("\nState Values:")
print("V(S0) =", round(V_S0, 2))
print("V(S1) =", round(V_S1, 2))

print("\nAction Values:")
print("Q(S0, Down) =", round(Q_S0_Down, 2))
print("Q(S0, Right) =", round(Q_S0_Right, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))
print("Q(S1, Up) =", round(Q_S1_Up, 2))

print("\nPolicy Consistency Checks:")

print("V(S0) = Q(S0, Down):",
      round(V_S0, 2) == round(Q_S0_Down, 2))

print("V(S1) = Q(S1, Right):",
      round(V_S1, 2) == round(Q_S1_Right, 2))
```
