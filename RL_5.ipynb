```python
import gymnasium as gym
import numpy as np


def random_agent(env_name, episodes=10):
    """
    Runs a Random Agent on the specified environment.

    Parameters:
        env_name: Name of the Gymnasium environment
        episodes: Number of episodes to run
    """

    env = gym.make(env_name)

    episode_rewards = []

    print("\n" + "=" * 50)
    print("Environment:", env_name)
    print("=" * 50)

    for episode in range(episodes):


        state, info = env.reset()

        total_reward = 0
        steps = 0

        while True:


            action_space = env.action_space


            action = action_space.sample()


            next_state, reward, terminated, truncated, info = env.step(action)


            print(
                f"Episode {episode + 1}, "
                f"Step {steps + 1}: "
                f"Action = {action}, "
                f"Reward = {reward}"
            )


            total_reward += reward
            steps += 1


            state = next_state


            if terminated or truncated:
                break

        episode_rewards.append(total_reward)

        print(
            f"Episode {episode + 1} completed | "
            f"Total Reward = {total_reward} | "
            f"Steps = {steps}"
        )


    average_reward = np.mean(episode_rewards)

    print("\nResults")
    print("-" * 50)
    print("Episode Rewards:", episode_rewards)
    print("Average Reward:", average_reward)

    env.close()

    return episode_rewards, average_reward


# -------------------------------------------------
# Run Random Agent on the three environments
# -------------------------------------------------

print("\nRANDOM AGENT - CARTPOLE")
random_agent("CartPole-v1", episodes=10)

print("\nRANDOM AGENT - MOUNTAINCAR")
random_agent("MountainCar-v0", episodes=10)

print("\nRANDOM AGENT - FROZENLAKE")
random_agent("FrozenLake-v1", episodes=10)
```
