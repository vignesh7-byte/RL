```python
n = int(input("Enter number of machines: "))

machines = []
rewards = []

for i in range(n):
    name = input(f"Enter machine name {i + 1}: ")
    r = list(map(float, input(f"Enter rewards for Machine {name}: ").split()))
    machines.append(name)
    rewards.append(r)

q = [0.0] * n
count = [0] * n

for i in range(n):
    print(f"\nMachine {machines[i]}")

    for r in rewards[i]:
        count[i] += 1
        q[i] = q[i] + (r - q[i]) / count[i]
        print(f"Reward: {r:g}, N({machines[i]}) = {count[i]}, Q({machines[i]}) = {q[i]:.2f}")

print("\nFinal Estimated Q Values")

for i in range(n):
    print(f"Machine {machines[i]}")
    print(f"Rewards: {[int(x) if x.is_integer() else x for x in rewards[i]]}")
    print(f"Q({machines[i]}) = {q[i]:.2f}")

selected = q.index(max(q))

print(f"\nGreedy Selected Machine: {machines[selected]}")
```
