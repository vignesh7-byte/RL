```python
%%writefile grid_env_det.py

# Grid World Environment

goal = (0, 2)
obstacle = (1, 1)

def transition(state, action):

    row, col = state

    if action == "up":
        row -= 1
    elif action == "down":
        row += 1
    elif action == "left":
        col -= 1
    elif action == "right":
        col += 1

    # Check grid boundaries
    if row < 0 or row > 1 or col < 0 or col > 2:
        return state

    # Check obstacle
    if (row, col) == obstacle:
        return state

    return (row, col)


def get_reward(state):

    if state == goal:
        return 10
    else:
        return -1
```

```python
import grid_env_det as env

gamma = 0.9

policy = {
    "S0": "right",
    "S1": "down",
    "S3": "right",
    "S4": "up"
}

states = {
    "S0": (0, 0),
    "S1": (0, 1),
    "S3": (1, 0),
    "S4": (1, 2)
}

goal = (0, 2)


def state_value_function(state):

    total_return = 0
    discount = 1
    current_state = states[state]

    for i in range(100):

        if current_state == goal:
            break

        action = policy[state]

        next_state = env.transition(current_state, action)
        reward = env.get_reward(next_state)

        total_return += discount * reward
        discount *= gamma

        current_state = next_state

        if current_state == goal:
            break

        for name, position in states.items():
            if position == current_state:
                state = name
                break

    return total_return


def action_value_function(state, action):

    total_return = 0
    discount = 1
    current_state = states[state]

    for i in range(100):

        if current_state == goal:
            break

        next_state = env.transition(current_state, action)
        reward = env.get_reward(next_state)

        total_return += discount * reward
        discount *= gamma

        current_state = next_state

        if current_state == goal:
            break

        for name, position in states.items():
            if position == current_state:
                state = name
                break

        action = policy[state]

    return total_return


print("State Value of S0:",
      state_value_function("S0"))

print("Action Value of S0 for UP:",
      action_value_function("S0", "up"))
```
