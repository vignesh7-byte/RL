```python

import gymnasium as gym

def random_agent(env_name, episodes=5):
    env = gym.make(env_name)

    print("\nEnvironment:", env_name)

    for episode in range(episodes):
        state, info = env.reset()

        total_reward = 0
        steps = 0
        terminated = False
        truncated = False

        while not terminated and not truncated and steps < 1000:
            action = env.action_space.sample()

            next_state, reward, terminated, truncated, info = env.step(action)

            total_reward += reward
            steps += 1
            state = next_state

        average_reward = total_reward / steps if steps > 0 else 0

        print("\nEpisode:", episode + 1)
        print("Total Reward:", total_reward)
        print("Number of Steps:", steps)
        print("Average Reward:", round(average_reward, 2))

    env.close()


random_agent("CliffWalking-v1", 5)
random_agent("Taxi-v3", 5)
```
